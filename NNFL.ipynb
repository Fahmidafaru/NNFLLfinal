```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('/content/owid-co2-data.csv')

# Numerical features select করি
numerical_features = ['gdp', 'population', 'cement_co2', 'coal_co2', 'gas_co2', 'oil_co2', 'co2', 'co2_per_gdp']
df[numerical_features] = df[numerical_features].fillna(0)

# Dummy images বানাই (64x64 grayscale)
dummy_image = np.random.rand(64, 64, 1)
df['image_data'] = [dummy_image for _ in range(len(df))]

# Target column
target_column = 'co2_per_capita'
df[target_column] = df[target_column].fillna(0)

# Final data
X_tabular = df[numerical_features].to_numpy()
X_images = np.stack(df['image_data'].values)
y = df[target_column].to_numpy()
```

```python
from sklearn.model_selection import train_test_split

X_img_train, X_img_test, X_tab_train, X_tab_test, y_train, y_test = train_test_split(
    X_images, X_tabular, y, test_size=0.2, random_state=42
)
```

```python
from tensorflow.keras import layers, models, Input, Model

# Image branch
image_input = Input(shape=(64, 64, 1))
x = layers.Conv2D(16, (3, 3), activation='relu')(image_input)
x = layers.MaxPooling2D((2, 2))(x)
x = layers.Conv2D(32, (3, 3), activation='relu')(x)
x = layers.Flatten()(x)

# Tabular branch
tabular_input = Input(shape=(len(numerical_features),))
y = layers.Dense(32, activation='relu')(tabular_input)

# Combine
combined = layers.concatenate([x, y])
z = layers.Dense(64, activation='relu')(combined)
z = layers.Dense(1)(z)  # Regression output

# Final model
model = Model(inputs=[image_input, tabular_input], outputs=z)
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()
```

```python
model.fit([X_img_train, X_tab_train], y_train, epochs=10, batch_size=32, validation_split=0.2)
```

```python
# Imports
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Concatenate, Flatten, Conv2D, MaxPooling2D
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Load CO₂ Dataset (tabular)
df = pd.read_csv('/content/owid-co2-data.csv')

# 2. Select numerical features (you can change as needed)
numerical_features = ['gdp', 'population', 'energy_per_capita']
target_column = 'co2_per_capita'

# Filter rows where target exists
df = df.dropna(subset=[target_column])

# Fill missing values in features
df[numerical_features] = df[numerical_features].fillna(0)

# Scale tabular data
scaler = StandardScaler()
X_tabular = scaler.fit_transform(df[numerical_features])
y = df[target_column].values

# 3. Generate dummy image data
num_samples = len(X_tabular)
X_images = np.random.randint(0, 255, size=(num_samples, 64, 64, 3)).astype('float32') / 255.0  # 64x64 to save RAM

# 4. Split train-test
X_tab_train, X_tab_test, X_img_train, X_img_test, y_train, y_test = train_test_split(
    X_tabular, X_images, y, test_size=0.2, random_state=42
)

# 5. Define Multi-Modal Model

# Image branch
image_input = Input(shape=(64, 64, 3))
x1 = Conv2D(16, (3,3), activation='relu')(image_input)
x1 = MaxPooling2D()(x1)
x1 = Conv2D(32, (3,3), activation='relu')(x1)
x1 = MaxPooling2D()(x1)
x1 = Flatten()(x1)

# Tabular branch
tabular_input = Input(shape=(X_tab_train.shape[1],))
x2 = Dense(64, activation='relu')(tabular_input)
x2 = Dense(32, activation='relu')(x2)

# Merge both
combined = Concatenate()([x1, x2])
z = Dense(64, activation='relu')(combined)
z = Dense(1)(z)  # Regression output

final_model = Model(inputs=[image_input, tabular_input], outputs=z)
final_model.compile(optimizer='adam', loss='mse', metrics=['mae'])

# 6. Train model
final_model.fit(
    [X_img_train, X_tab_train],
    y_train,
    validation_split=0.2,
    epochs=10,
    batch_size=32
)

# 7. Evaluate
loss, mae = final_model.evaluate([X_img_test, X_tab_test], y_test)
print(f"Test MAE: {mae:.4f}")
```
